## Latar Belakang Bisnis

Indonesia adalah negara agraris yang sangat bergantung pada pertanian. Nusa Tenggara Barat, khususnya Kabupaten Sumbawa, memiliki peran penting dalam produksi tanaman pangan. Data BPS Provinsi NTB mencatat 77.496 unit rumah tangga usaha pertanian pada tahun 2023, dan komoditas seperti beras, jagung, dan kacang hijau menjadi sumber pendapatan utama bagi masyarakat lokal.

## Problem Statement

Fluktuasi harga komoditas di Kabupaten Sumbawa menciptakan ketidakstabilan ekonomi bagi petani, pedagang, dan pemerintah daerah. Salah satu contoh nyata adalah penurunan harga jagung hingga Rp 3.800/kg pada 29 April 2024, yang menyebabkan kerugian petani dan unjuk rasa publik (Gustiana, 2024). Proyek ini bertujuan mengurangi ketidakpastian tersebut dengan memberikan prediksi harga yang lebih akurat.

## Objective

- Mengembangkan prediksi harga untuk empat seri komoditas utama: Beras Premium, Beras Medium, Jagung Pipil Kering, dan Kacang Hijau.
- Membangun empat model LSTM univariat terpisah, masing-masing untuk setiap komoditas.
- Mengembangkan satu model LSTM multivariat sebagai benchmark.
- Menggunakan data historis harga dan variabel eksogen seperti produktivitas, musim, dan kalender libur.
- Mendukung keputusan operasional petani, pedagang, dan pemerintah daerah.

## Scope Proyek

Dataset saat ini mencakup empat seri harga komoditas utama:
- Beras Premium
- Beras Medium
- Jagung Pipil Kering
- Kacang Hijau

Selain data harga harian, tersedia juga dataset produktivitas triwulanan yang dapat digunakan sebagai fitur eksogen dalam model.

    Business Understanding   - Define objectives & scope
    Data Understanding       - Explore raw datasets
    Data Preparation         - Clean, aggregate, feature engineering
    Modeling                 - Train LSTM univariate & multivariate
    Evaluation               - Validate and compare metrics

## Model Strategy

Strategi pemodelan yang akan digunakan pada fase berikutnya adalah:
1. Empat model LSTM univariat terpisah untuk masing-masing komoditas.
2. Satu model LSTM multivariat sebagai benchmark untuk menangkap hubungan lintas komoditas.

Pendekatan univariat memungkinkan setiap model mempelajari pola harga khusus komoditas, sedangkan model multivariat mengeksplorasi korelasi antar harga komoditas.

## Data Mentah yang Tersedia

- data/raw/DATASET HARGA KOMODITI.csv: data harga komoditas harian dari 2022-01-01 hingga 2024-11-30.
- data/raw/produksi.csv: data produktivitas dan produksi triwulanan.

Kolom harga utama meliputi Komoditi, Tanggal, Harga Petani, dan Harga Pengecer. Data produktivitas meliputi Luas Panen, Produktivitas, dan Produksi.

## Pemangku Kepentingan

- Petani: membutuhkan prediksi harga untuk merencanakan produksi dan penjualan.
- Pedagang: memerlukan sinyal harga untuk mengelola persediaan dan distribusi.
- Pemerintah Daerah: memerlukan informasi untuk kebijakan stabilisasi harga dan intervensi pasar.
- Tim Data Science: bertanggung jawab menyediakan model yang dapat divalidasi, direproduksi, dan diterapkan secara operasional.

## Expected Benefits

- Peningkatan reliabilitas prediksi harga komoditas.
- Dukungan bagi pengambilan keputusan produksi dan distribusi.
- Dasar bagi kebijakan stabilisasi harga oleh pemerintah daerah.
- Kerangka kerja yang siap dikembangkan ke tahap Data Understanding dan Modeling.
